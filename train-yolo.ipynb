```python
from ultralytics import YOLO
import torch
import os
```

```python
# ============================================================
# 1. CHECK DEVICE
# ============================================================

print("=" * 60)
print("YOLO26 OBJECT DETECTION TRAINING")
print("=" * 60)

if torch.cuda.is_available():
    device = 0
    print("GPU available:")
    print(torch.cuda.get_device_name(0))
else:
    device = "cpu"
    print("CUDA GPU not available. Using CPU.")
```

```python
DATASET = "dataset/data.yaml"
```

```python
# load pretrained YOLO26 model
model = YOLO("yolo26n.pt")  # load a pretrained YOLO
```

```python
# train model
results = model.train(
    # Dataset
    data=DATASET,
    # Training
    epochs=100,
    imgsz=640,
    # Hardware
    device=device,
    # Batch
    batch=16,
    # Workers
    workers=4,
    # Output
    project="runs/yolo26",
    name="custom_object_detection",
    # Save
    save=True,
    save_period=10,
    # Validation
    val=True,
    # Early stopping
    patience=20,
    # Augmentation
    mosaic=1.0,
    mixup=0.0,
    # Reproducibility
    seed=42,
    # Cache
    cache=False,
    # Plot training results
    plots=True,
    # Verbose output
    verbose=True
)

print("\n" + "=" * 60)
print("TRAINING COMPLETE")
print("=" * 60)
print("Results saved in:")
print("runs/yolo26/custom_object_detection/")
print("\nBest model:")
print(
    "runs/yolo26/custom_object_detection/weights/best.pt"
)
print("\nLast model:")
print(
    "runs/yolo26/custom_object_detection/weights/last.pt"
)
```
